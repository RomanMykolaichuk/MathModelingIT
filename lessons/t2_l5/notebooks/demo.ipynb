# T2.L5 Demo — множинний вибір
WSM, TOPSIS, sensitivity та robustness.

```
from pathlib import Path
import sys, pandas as pd, matplotlib.pyplot as plt
BASE = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(BASE/"src"))
from model import weighted_sum, topsis, one_factor_sensitivity, robustness_analysis
M = pd.read_csv(BASE/"data"/"alternatives.csv").set_index("alternative")
C = pd.read_csv(BASE/"data"/"criteria.csv")
W = C.set_index("criterion")["weight"].astype(float)
T = C.set_index("criterion")["type"].to_dict()
M, W
```

## 1. Baseline WSM

```
wsm = weighted_sum(M,W,T)
print(wsm.ranking)
display(wsm.scores.sort_values(ascending=False).to_frame("WSM score"))
```

## 2. Baseline TOPSIS

```
top = topsis(M,W,T)
print(top.ranking)
display(top.scores.sort_values(ascending=False).to_frame("TOPSIS score"))
```

## 3. Порівняння

```
cmp = pd.DataFrame({"WSM":wsm.scores,"TOPSIS":top.scores})
display(cmp)
cmp.plot(kind="bar", title="WSM vs TOPSIS")
plt.ylabel("Нормований бал")
plt.show()
```

## 4. Sensitivity до ваги reliability

```
sens = one_factor_sensitivity(M,W,T,"reliability",[0.10,0.15,0.20,0.25,0.30,0.35,0.40,0.45])
display(sens)
```

## 5. Robustness analysis

```
rob = robustness_analysis(M,W,T,n_runs=3000,concentration=80,seed=2026)
pivot = rob.pivot(index="alternative",columns="method",values="top_choice_share")
display(pivot)
pivot.plot(kind="bar", title="Частка першого місця при варіюванні ваг")
plt.ylabel("Top-choice share")
plt.show()
```

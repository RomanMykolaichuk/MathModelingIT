## 6. Аналіз чутливості

```python
# TODO: використайте не менше п'яти значень rate
rates = [4, 6, 8, 10, 12]
sensitivity = compare_rates(times, s0=S0, rates=rates)
sensitivity.head()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
for r, group in sensitivity.groupby('scenario_rate'):
    ax.plot(group['time'], group['resource'], label=f'rate={r:g}')
ax.set_title('Чутливість до параметра rate')
ax.set_xlabel('Час, умовні одиниці')
ax.set_ylabel('Залишок ресурсу, умовні одиниці')
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

## 5. Аналіз чутливості до бюджету

Дослідницьке питання: чи кожна додаткова одиниця бюджету однаково покращує оптимальний результат?


```python
sensitivity = run_budget_sensitivity()
sensitivity
```

```python
valid = sensitivity[sensitivity["success"]]
ax = valid.plot(
    x="budget",
    y="objective",
    marker="o",
    legend=False,
    figsize=(8, 5),
    title="Чутливість оптимального результату до бюджету",
)
ax.set_xlabel("Доступний бюджет")
ax.set_ylabel("Значення цільової функції")
plt.show()
```
